9. Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. С помощью `while` посчитайте, через сколько месяцев он накопит. А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?

````python
#6
rev_almaty = 0
rev_astana = 0
for c, a, p in zip(ss, ff, paid):
    if p and c == "Almaty":
        rev_almaty += a
    elif p and c == "Astana":
        rev_astana += a

print(f"Алматы: {rev_almaty:,} ₸")
print(f"Астана: {rev_astana:,} ₸")
````

````python
#7
max_idx = -1
max_amount = 0
for i, a in enumerate(ff):
    if paid[i] and a > max_amount:
        max_amount = a
        max_idx = i
print(f"Индекс: {max_idx}, город: {ss[max_idx]}, сумма: {max_amount:,} ₸")
````
